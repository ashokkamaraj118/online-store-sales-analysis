# DATA CLEANING

```python
df.isnull().sum()
```

```python
df=df.dropna()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["City"].unique()
```

```python
df["City"] = df["City"].replace({
    "Chennai":"madras",
    "Banglore":"karnataka","Ahmedabad":"Gujarat"
})
```

```python
df.info
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.rename(columns={'Citys': 'Date'})
```

```python
df.info
```

# 3 Data processing

```python
df["Revenue"] = df["Quantity"] * df["Price"]
```

```python
df.head()
```

```python
total_revenue = df["Revenue"].sum()

print("Total Revenue =", total_revenue)
```

```python
best_product = df.groupby("Product")["Quantity"].sum()

best_product
```

```python
best_product.idxmax()
```

```python
best_product.max()
```

```python
print("Best Selling Product:", best_product.idxmax())
print("Quantity Sold:", best_product.max())
```

```python
city_revenue = df.groupby("City")["Revenue"].sum()

city_revenue
```

```python
print("Best City:", city_revenue.idxmin())
print("Revenue:", city_revenue.min())
```

```python
df.info
```

```python
category_revenue = df.groupby("Category")["Revenue"].sum()

category_revenue
```

```python
df["Month"] = df["Date"].dt.month_name()
```

```python
df.head()
```

```python
monthly_revenue =df.groupby("Month")["Revenue"].sum()

monthly_revenue
```

```python
month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]
```

```python
monthly_revenue = monthly_revenue.reindex(month_order)

monthly_revenue
```

```python
city_revenue = df.groupby("City")["Revenue"].sum()

plt.figure(figsize=(12,12))

plt.bar(city_revenue.index, city_revenue.values)

plt.title("Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue")

plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.pie(category_revenue,
        labels=category_revenue.index,
        autopct="%1.1f%%")

plt.title("Revenue Contribution by Category")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(monthly_revenue.index,
         monthly_revenue.values,
         marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")


plt.grid(True)

plt.show()
```
