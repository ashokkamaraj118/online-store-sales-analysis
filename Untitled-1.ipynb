```python
# Highest quantity order
highest_quantity = df.loc[df["Order Quantity"].idxmax()]

print(highest_quantity)
```

```python
print("Highest Revenue Order")
print("Order No:", highest_quantity["Order No"])
print("Customer:", highest_quantity["Customer Name"])
print("Revenue:", highest_quantity["Revenue"])
```

```python
# Maximum quantity value
df["Order Quantity"].max()
```

```python
# Row index of maximum quantity
df["Order Quantity"].idxmax()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
# Create Month column
df["Month"] = df["Order Date"].dt.month_name()
```

```python
# Monthly revenue
monthly_sales = df.groupby("Month")["Revenue"].sum()

print(monthly_sales)
```

```python
print("Best Sales Month:", monthly_sales.idxmax())
print("Revenue:", monthly_sales.max())
```

```python
total_tax = df["Tax (USD)"].sum()

print("Total Tax Collected: $", total_tax)
```

```python
monthly_sales=df.groupby("Month")["Revenue"].sum()
plt.Figure (figsize=(8,5))
plt.bar(monthly_sales.index,monthly_sales.values)
plt.title("bar chart")
plt.xlabel("month")
plt.ylabel("revenue")
plt.show

```

```python
values = [df["Revenue"].sum(), df["Tax (USD)"].sum()]
labels = ["Revenue", "Tax"]
plt.Figure(figsize=(8,5))
plt.pie(values,
        labels=labels,
        autopct="%1.1f%%",
        startangle=90)
plt.title("revenue vs tax")
plt.show
```
